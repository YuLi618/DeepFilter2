```python
# Sydney Housing Project - V1.0 Data Ingestion & Basic EDA

from clearml import Task
import pandas as pd

# 1. Initialize ClearML tracker (Automatically logs all console outputs and data flows)
task = Task.init(project_name="Sydney_Housing_Project", task_name="V1_Data_Ingestion_and_EDA")

print(" Connecting to ClearML and loading NSW rental bond data...")

# 2. Load data (Ensure the file is renamed to nsw_rental_bonds.csv)
from pathlib import Path

BASE_DIR = Path.cwd()

# If ClearML agent runs from a different folder, search upward for the dataset file.
for parent in [BASE_DIR] + list(BASE_DIR.parents):
    candidate = parent / "nsw_rental_bonds.csv"
    if candidate.exists():
        file_path = candidate
        break
else:
    raise FileNotFoundError("Could not find nsw_rental_bonds.csv")

df = pd.read_csv(file_path)
print(f"Loaded dataset from: {file_path}")

# 3. Basic architecture review (Check the shape and features of our dataset)
print("\n Data loaded successfully!")
print(f" Dataset shape: {df.shape[0]} samples (rows), {df.shape[1]} features (columns)")

# 4. Preview the first 5 rows (Jupyter's display function renders a nice table)
print("\n Core data preview:")
display(df.head())

# 5. Scan data health (Find missing values, the ultimate algorithm killer)
print("\n Missing values (NaN) radar:")
# Only display columns that contain missing values
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])

# 6. Explore our target variable (e.g., Rent distribution)
# Note: You need to replace 'Rent' with the actual column name for rent in the dataset
# print("\n Rent distribution summary:")
# print(df['Rent'].describe())
task.close()
```
